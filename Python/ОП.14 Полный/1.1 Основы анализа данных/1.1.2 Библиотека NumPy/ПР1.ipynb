## Задание 5. Операции между массивами
1. Создайте копию вашего двухмерного массива `n2` с помощью метода `.copy()`.
2. Создайте новый двухмерный массив такого же размера.
3. Выполните с ними математические операции: сложение, вычитание, умножение и деление массивов.

```python
def copy_math(x):

    x_copy = np.copy(x)

    x_work = np.array([[1, 2 ,3 ],
                          [4, 8 ,9]])

    print("Вычитание:\n", x_copy - x_work)
    print("Сложение:\n", x_copy + x_work)
    print("Умножение:\n", x_copy * x_work)
    print("Деление:\n", x_copy / x_work)

n2 = np.array([[5, 10, 15],
               [20, 25, 30]])
copy_math(n2)
```

## Задание 7. Тригонометрические функции
Создайте новый массив из случайных чисел и вычислите для его элементов косинус, синус и тангенс.

```python
# Ваш код здесь

def trig (x):
    angles_rad = np.radians(x)
    angels_sin = np.sin(angles_rad)
    angels_con = np.cos(angles_rad)
    anglets_tg = np.tan(angles_rad)

    print("Радианы:\n", angles_rad)
    print("Синусы:\n", angels_sin)
    print("Косинусы:\n", angels_con)
    print("Тангенсы:\n", anglets_tg )

angles_deg = np.array([0, 30, 45, 60, 90])
trig(angles_deg)
```

## Задание 8. Булевы выражения
Примените два любых булевых выражения к массиву из случайных чисел.

```python
rng = np.random.default_rng()

df_rng = rng.integers(low = 10, high = 100, size = (3, 4))

mask1 = (df_rng > 15)
mask2 = (df_rng < 30)
print(mask1)
print(mask2)
```

## Задание 9. Работа со срезами (Slicing)
Поэкспериментируйте со срезами: выведите первую строку массива, первый столбец массива и выбранный диапазон.

```python
# Ваш код здесь
x_work = np.array([[1, 2 ,3 ],
                    [4, 8 ,9]])

first = x_work[:1, :]
second = x_work[:, 0]

diap = x_work[:5]

print(first)
print(second)
print(diap)
```
