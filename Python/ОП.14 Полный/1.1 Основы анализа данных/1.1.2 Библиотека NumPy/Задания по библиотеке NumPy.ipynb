## Задание 3: Продвинутые операции с массивами NumPy

### 1. Матричные операции.
Создайте две матрицы заданной размерности.
Выполняйте сложение, вычитание и умножение матриц.

```python
data1 = np.random.randn(10, 3)  # Generating random data with 100 rows and 3 columns
data2 = np.random.randn(10, 3)  # Generating another set of random data with the same shape

summ = data1 + data2  # Element-wise addition of the two datasets
razn = data1 - data2  # Element-wise subtraction of the two datasets
multiplication = data1 * data2  # Element-wise multiplication of the two datasets

print("Сумма:\n", summ)
print("Разность:\n", razn)
print("Произведение:\n", multiplication)
```

### 2. Булевы операции.
Примените булевы операции к массиву.

```python
np.random.seed(42)  # Setting a seed for reproducibility

normal_dist = np.random.normal(loc=0, scale=1, size=1000)  # Generating 1000 samples from a normal distribution
uniform_dist = np.random.uniform(low=0, high=1, size=1000)  # Generating 1000 samples from a uniform distribution

random_matrix = np.random.rand(5, 5)  # Generating a 5x5 matrix of random numbers from a uniform distribution

arr = np.arange(10)
np.random.shuffle(arr)  # Shuffling the array in place

print("Shuffled array:", arr)
print("Random normal distribution samples:\n", normal_dist[:10])  # Print first 10 samples
print("Random uniform distribution samples:\n", uniform_dist[:10])  # Print first
print("Random matrix:\n", random_matrix)
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

C = np.dot(A, B)  # Matrix multiplication using np.dot

print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("Matrix C (A dot B):\n", C)

eingenvalues, eigenvectors = np.linalg.eig(A)  # Computing eigenvalues and eigenvectors of matrix A
print("Eigenvalues of A:", eingenvalues)
print("Eigenvectors of A:\n", eigenvectors)
```

```python
# Пример нормализации данных

def normalize_features(X):
    mean = np.mean(X, axis=0)  # Compute the mean of each feature
    std = np.std(X, axis=0)  # Compute the standard deviation of each
    return (X - mean) / std  # Normalize the features


X = np.random.rand(100, 3)  # Generating a random dataset with 100 samples and 3 features
X_normalized = normalize_features(X)  # Normalize the features

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))  # Compute cosine similarity between two vectors


v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
similarity = cosine_similarity(v1, v2)  # Calculate cosine similarity between v1 and v2
print("Cosine similarity between v1 and v2:", similarity)
```
