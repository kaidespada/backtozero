## Задание №1

1. Прежде чем заполнять пропуски, их нужно найти. Напечатайте на экране строки таблицы, у которых есть пропуски в столбце `'faculty_name'`. Данные находятся в файле `hogwarts_points.csv`.

```python
import pandas as pd

# Аргумент sep нужен для указания символа разделителя в файле данных. По умолчанию это ,
# Но бывают и дургие разделители, например ;
hog = pd.read_csv('hogwarts_points.csv', sep=';')

hog
```

2. Восстановите справедливость: верните Рону его факультет, заменив значение `NaN` на `'Гриффиндор'` методом `fillna()`. Выведите на экран таблицу `hogwarts_points`. Данные находятся в файле `hogwarts_points.csv`.

```python
hog = hog.fillna('Гриффиндор')

hog
```

3. Найдите суммарные баллы каждого из факультетов и сохраните результат в переменной `faculty_points`. Потом используя эту переменную найдите факультет-победитель и сохраните результат в переменной `winner`. В конце выведите значения обеих переменных на экран. Данные находятся в файле `hogwarts_points.csv`.

```python
faculty_point = hog.groupby('faculty_name')['points'].sum()

winner = hog.groupby('faculty_name')['points'].sum().idxmax()

print(faculty_point)

print()

print(winner)
```

4. Применим полученные в Хогвартсе знания к реальной задаче. Возьмём источники трафика Яндекс.Маркета и рассчитаем конверсию.

5. Посчитайте количество совершённых покупок для каждого источника трафика. Результат сохраните в переменной `purchase` и выведите на экран. Данные находятся в файле `logs.csv`.

```python
df = pd.read_csv('logs.csv')
purchase = df.groupby('source')['purchase'].sum()

purchase
```

6. Посчитайте конверсию по каждому источнику трафика, результат сохраните в переменной `conversion` и выведите на экран. Данные находятся в файле `logs.csv`.
