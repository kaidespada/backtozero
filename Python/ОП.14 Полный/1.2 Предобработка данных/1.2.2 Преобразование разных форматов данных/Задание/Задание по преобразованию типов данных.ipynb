### Задание 1: Преобразование строковых дат в даты
**Задача:** Преобразовать столбец дат в строковом формате (YYYY-MM-DD) в объекты времени в Pandas.

**Набор данных:** Данные о продажах (`sales_data.csv`) со столбцами для идентификатора продажи, продукта и даты продажи (в строковом формате).

**Задачи:**

1. Загрузите набор данных в Pandas DataFrame.
2. Преобразуйте столбец "Дата продажи" из строкового формата в объект datetime.
3. Извлеките год, месяц и день в виде отдельных столбцов.
4. отфильтруйте набор данных, чтобы включить в него только продажи за текущий год.

```python
import pandas as pd

df = pd.read_csv("sales_data.csv")
df['Sale Date'] = df['Sale Date'].astype('datetime64[s]')
df['Year'] = df['Sale Date'].dt.year
df['Month'] = df['Sale Date'].dt.month
df['Day'] = df['Sale Date'].dt.day

df_sort = df[df['Year'] >= 2024]
df_sort
```
