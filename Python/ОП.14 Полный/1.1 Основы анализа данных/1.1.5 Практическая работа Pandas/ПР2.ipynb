### Таск 5. Какую долю от всех покупателей составляют вместе мужчины от 26 до 35 лет и женщины старше 36 лет?

```python
# 1. Фильтруем данные (указываем точные текстовые диапазоны для возраста)
typedata = df[((df['Gender'] == 'M') & (df['Age'] == '26-35')) | 
              ((df['Gender'] == 'F') & (df['Age'].isin(['36-45', '46+'])))]

# 2. Считаем долю через длину датафреймов (так как typedata — это уже отфильтрованная таблица)
share = len(typedata) / len(df)
percentage = share * 100

# 3. Выводим результат (исправил также букву f перед строкой в вашем принте)
print(f"Доля: {share:.4f}")
print(f"В процентах: {percentage:.2f}%")
```
